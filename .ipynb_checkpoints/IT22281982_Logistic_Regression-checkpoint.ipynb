```python
# Load dataset (make sure Telco_csv.csv is in the same folder as this notebook)
df = pd.read_csv('Telco_csv.csv')

print('Dataset Shape:', df.shape)
print('\nFirst 5 rows:')
df.head()
```

```python
# Check churn distribution
print('Churn value counts:')
print(df['Churn'].value_counts())

plt.figure(figsize=(5, 4))
df['Churn'].value_counts().plot(kind='bar', color=['steelblue', 'tomato'], edgecolor='black')
plt.title('Churn Distribution')
plt.xlabel('Churn')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('lr_churn_distribution.png', dpi=150)
plt.show()
print('Saved: lr_churn_distribution.png')
```
